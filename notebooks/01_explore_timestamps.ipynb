```python
import pandas as pd
from pathlib import Path
```

```python
raw_data = Path("..") / "data" / "raw"
solar_data = raw_data / "solar"

p1_gen = pd.read_csv(solar_data / "Plant_1_Generation_Data.csv")
p2_gen = pd.read_csv(solar_data / "Plant_2_Generation_Data.csv")

p1_weather = pd.read_csv(solar_data / "Plant_1_Weather_Sensor_Data.csv")
p2_weather = pd.read_csv(solar_data / "Plant_2_Weather_Sensor_Data.csv")

p1_gen.head()
```

Clean Time series (Date-Time)

```python
#Generation Data
p1_gen["DATE_TIME"] = pd.to_datetime(p1_gen["DATE_TIME"], format="%d-%m-%Y %H:%M")
p2_gen["DATE_TIME"] = pd.to_datetime(p2_gen["DATE_TIME"], format="%Y-%m-%d %H:%M:%S")

#Weather Data
p1_weather["DATE_TIME"] = pd.to_datetime(p1_weather["DATE_TIME"], format="%Y-%m-%d %H:%M:%S")
p2_weather["DATE_TIME"] = pd.to_datetime(p2_weather["DATE_TIME"], format="%Y-%m-%d %H:%M:%S")
```

Check whether the time range in files are equal

```python
if p1_gen["DATE_TIME"].min() == p1_weather["DATE_TIME"].min():
    if p1_gen["DATE_TIME"].max() == p1_weather["DATE_TIME"].max():
        if p2_gen["DATE_TIME"].min() == p2_weather["DATE_TIME"].min():
            if p2_gen["DATE_TIME"].max() == p2_weather["DATE_TIME"].max():
                print ("Time range match check pass")
            else:
              print ("Plant 2 Generation and weather data end date mismatch")  
        else:
            print ("Plant 2 Generation and weather data start date mismatch")
    else:
        print ("Plant 1 Generation and weather data end date mismatch")
else:
    print ("Plant 1 Generation and weather data start date mismatch")

print("Time series Start at:", p1_gen["DATE_TIME"].min())
print("Time series end at:", p1_gen["DATE_TIME"].max())
```
